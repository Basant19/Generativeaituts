```python
#structured output approach using typed dict


from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")

llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    google_api_key=api_key
)

from typing import TypedDict

class Explanation(TypedDict):
    topic: str
    summary: str
    difficulty: str

structured_llm = llm.with_structured_output(Explanation)
prompt="Explain Binary Search for beginners."
response = structured_llm.invoke(prompt)

print(response)

#print(type(response))

```

```python
# output structured using jason schema
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv          
import os
load_dotenv()   
api_key = os.getenv("GOOGLE_API_KEY")
llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    google_api_key=api_key
)
json_schema = {
    "type": "object",
    "properties": {
        "topic": {"type": "string"},
        "summary": {"type": "string"},
        "difficulty": {"type": "string"}
    },
    "required": ["topic", "summary", "difficulty"]
}
structured_llm = llm.with_structured_output(json_schema)
prompt = "Explain Merge Sort algorithm for beginners."
response = structured_llm.invoke(prompt)
print(response)
```

```python
# example 2 : runnable lambda function 

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnableLambda
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import os

# Load API key
load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")

# Gemini LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    temperature=0.5,
    google_api_key=api_key
)

# Normal Python function
def build_prompt(topic: str) -> str:
    return f"""
You are a teacher AI.
Explain the topic '{topic}' in very simple language.
Give a real-life example.
"""

# Convert function → RunnableLambda
prompt_builder = RunnableLambda(build_prompt)

# Output parser
parser = StrOutputParser()

# Build chain
chain = prompt_builder | llm | parser

# Invoke
response = chain.invoke("Hash Table")

print(response)

```

```python
# example 3 : runnable lambda function 
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnableLambda
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import os

# Load API key
load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")

# Initialize Gemini LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    temperature=0.5,
    google_api_key=api_key
)

# RunnableLambda: preprocess input
preprocess = RunnableLambda(
    lambda topic: f"Explain {topic} in simple terms."
)

# Output parser
parser = StrOutputParser()

# Build chain
chain = preprocess | llm | parser

# Invoke chain
response = chain.invoke("Binary Search")

print(response)
```

```python
# example runnable parallel
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnableParallel, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import os

# Load env
load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")

# Gemini LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    temperature=0.3,
    google_api_key=api_key
)

parser = StrOutputParser()

# ---- Runnable functions ----

explain = RunnableLambda(
    lambda topic: f"Explain {topic} in simple terms."
)

complexity = RunnableLambda(
    lambda topic: f"What is the time and space complexity of {topic}?"
)

example = RunnableLambda(
    lambda topic: f"Give a real-world example of {topic}."
)

# ---- Parallel execution ----
parallel_chain = RunnableParallel(
    explanation=explain | llm | parser,
    complexity=complexity | llm | parser,
    example=example | llm | parser
)

# Invoke
result = parallel_chain.invoke("Binary Search")

# Output
for key, value in result.items():
    print(f"\n🔹 {key.upper()}:\n{value}")
```

```python
#runnable conditional example
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnableBranch, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import os

# Load env
load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")

# LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    temperature=0.3,
    google_api_key=api_key
)

parser = StrOutputParser()

# ---- Prompt creators ----
code_prompt = RunnableLambda(
    lambda topic: f"Explain {topic} with code examples."
)

math_prompt = RunnableLambda(
    lambda topic: f"Explain {topic} with mathematical details."
)

simple_prompt = RunnableLambda(
    lambda topic: f"Explain {topic} in simple terms."
)

# ---- Conditional Branch ----
conditional_chain = RunnableBranch(
    (
        lambda x: "code" in x.lower(),
        code_prompt | llm | parser
    ),
    (
        lambda x: "math" in x.lower(),
        math_prompt | llm | parser
    ),
    # default branch (else)
    simple_prompt | llm | parser
)

# ---- Invoke ----
response = conditional_chain.invoke("Binary Search with code")

print(response)
```
